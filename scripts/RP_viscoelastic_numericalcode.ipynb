## 1. Governing Equations and Integral Formulation

The radial momentum balance for an incompressible viscoelastic fluid surrounding a pulsating spherical bubble yields the Generalized Rayleigh–Plesset equation:

$$
\rho \left( R \ddot{R} + \frac{3}{2} \dot{R}^2 \right) = \Delta p - \frac{2\gamma}{R} + \int_{R}^{\infty} \left( 2\sigma_{rr} - \sigma_{\theta\theta} - \sigma_{\phi\phi} \right) \frac{\mathrm{d}r}{r} \tag{1}
$$

Because bubble pulsation is spherically symmetric, the surrounding liquid expands and contracts equally in all tangential directions. Consequently, the azimuthal stress $\sigma_{\theta\theta}$ and polar stress $\sigma_{\phi\phi}$ are equal ($\sigma_{\theta\theta} = \sigma_{\phi\phi}$), simplifying the integrand in (1) to:

$$
\int_{R}^{\infty} \left( 2\sigma_{rr} - \sigma_{\theta\theta} - \sigma_{\phi\phi} \right) \frac{\mathrm{d}r}{r} = \int_{R}^{\infty} \frac{2(\sigma_{rr} - \sigma_{\theta\theta})}{r} \, \mathrm{d}r \tag{2}
$$

Here, $R(t)$ denotes the instantaneous bubble radius, overdots represent temporal differentiation, $\rho$ is the fluid density, $\gamma$ is the interfacial tension coefficient, and $\Delta p = P_g(R) - P_0 - P_A(t)$ represents the dynamic pressure differential between the internal gas pressure $P_g(R)$, the static ambient pressure $P_0$, and the driving acoustic pressure $P_A(t) = -p_A \sin(2\pi f t)$ with amplitude $p_A$ and frequency $f$.

Assuming polytropic behavior for the enclosed non-condensable gas, the internal gas pressure takes the form

$$
P_g(R) = \left(P_0 + \frac{2\gamma}{R_0}\right) \left(\frac{R_0}{R}\right)^{3\kappa} \tag{3}
$$

with $R_0$ denoting the equilibrium bubble radius and $\kappa$ the polytropic exponent. The quantities $\sigma_{rr}$ and $\sigma_{\theta\theta}$ represent the radial and azimuthal components of the deviatoric stress tensor $\boldsymbol{\sigma}$.

For an Upper Convected Maxwell (UCM) fluid, the stress tensor $\boldsymbol{\sigma}(t)$ is expressed as a hereditary integral over past deformation histories weighted by an exponential relaxation kernel $\psi(t)$:

$$
\boldsymbol{\sigma}(t) = \int_{-\infty}^{t} \psi(t - t') \mathbf{C}_t^{-1}(t') \, \mathrm{d}t' \tag{4}
$$

where $\mathbf{C}_t^{-1}(t')$ is the Finger strain tensor relating the past state at time $t'$ to the current state at time $t$. The relaxation kernel is defined as

$$
\psi(t) = \frac{\eta}{\lambda} \exp\left(-\frac{t}{\lambda}\right) = G \exp\left(-\frac{t}{\lambda}\right) \tag{5}
$$

where $\eta$ is the zero-shear dynamic viscosity ($\eta = 0.03\,\mathrm{Pa\,s}$), $\lambda = \mathrm{De}/(2\pi f)$ is the relaxation time associated with the Deborah number $\mathrm{De}$, and $G = \eta/\lambda$ denotes the elastic shear modulus.

---

## 2. Transition from Integral to Differential Formulation

Evaluating the memory integral at every time step requires storing the complete deformation history, scaling computationally as $\mathcal{O}(N^2)$ and encountering severe numerical instabilities during rapid, large-amplitude bubble collapses. To overcome these limitations, the integral UCM relation is transformed into its frame-invariant differential representation:

$$
\boldsymbol{\sigma} + \lambda \stackrel{\nabla}{\boldsymbol{\sigma}} = 2 \eta \mathbf{D} \tag{6}
$$

where $\mathbf{D} = \frac{1}{2}\left[ \nabla \mathbf{v} + (\nabla \mathbf{v})^T \right]$ is the rate-of-deformation tensor, and $\stackrel{\nabla}{\boldsymbol{\sigma}}$ represents the upper-convected objective rate, defined as

$$
\stackrel{\nabla}{\boldsymbol{\sigma}} = \frac{\mathrm{D}\boldsymbol{\sigma}}{\mathrm{D}t} - (\nabla \mathbf{v})^T \cdot \boldsymbol{\sigma} - \boldsymbol{\sigma} \cdot (\nabla \mathbf{v}) \tag{7}
$$

---

## 3. Kinematics and Stress Components at the Interface

Under the assumptions of fluid incompressibility and spherical symmetry, the radial velocity field $u(r,t)$ surrounding the cavity is expressed as

$$
u(r, t) = \frac{R^2 \dot{R}}{r^2} \tag{8}
$$

Evaluating the velocity gradient tensor $\nabla \mathbf{v}$ at the bubble interface ($r = R$) yields

$$
\left.\nabla \mathbf{v}\right|_{r=R} = \begin{pmatrix} \frac{\partial u}{\partial r} & 0 & 0 \\ 0 & \frac{u}{r} & 0 \\ 0 & 0 & \frac{u}{r} \end{pmatrix}_{r=R} = \begin{pmatrix} -2\frac{\dot{R}}{R} & 0 & 0 \\ 0 & \frac{\dot{R}}{R} & 0 \\ 0 & 0 & \frac{\dot{R}}{R} \end{pmatrix} \tag{9}
$$

Defining the interfacial elongation rate as $\dot{\varepsilon} = \dot{R}/R$, the non-zero components of the rate-of-deformation tensor become $D_{rr} = -2\dot{\varepsilon}$ and $D_{\theta\theta} = D_{\phi\phi} = \dot{\varepsilon}$. Substituting these kinematic conditions into (7), the components of the upper-convected derivative evaluate to

$$
\stackrel{\nabla}{\sigma}_{rr} = \frac{\mathrm{d}\sigma_{rr}}{\mathrm{d}t} + 4\dot{\varepsilon}\sigma_{rr} \tag{10}
$$

$$
\stackrel{\nabla}{\sigma}_{\theta\theta} = \frac{\mathrm{d}\sigma_{\theta\theta}}{\mathrm{d}t} - 2\dot{\varepsilon}\sigma_{\theta\theta} \tag{11}
$$

---

## 4. Derivation of the Differential Governing Equations

Substituting the frame-invariant derivatives (10)–(11) and strain-rate components into the UCM constitutive relation (6) isolates the explicit time derivatives of the radial and azimuthal stresses:

$$
\frac{\mathrm{d}\sigma_{rr}}{\mathrm{d}t} = -4G\left(\frac{\dot{R}}{R}\right) - \frac{\sigma_{rr}}{\lambda} - 4\left(\frac{\dot{R}}{R}\right)\sigma_{rr} \tag{12}
$$

$$
\frac{\mathrm{d}\sigma_{\theta\theta}}{\mathrm{d}t} = 2G\left(\frac{\dot{R}}{R}\right) - \frac{\sigma_{\theta\theta}}{\lambda} + 2\left(\frac{\dot{R}}{R}\right)\sigma_{\theta\theta} \tag{13}
$$

Performing the spatial integration of the normal stress difference across the liquid domain reduces the integral term in (2) to interface-evaluated quantities:

$$
\int_{R}^{\infty} \frac{2(\sigma_{rr} - \sigma_{\theta\theta})}{r} \, \mathrm{d}r = \sigma_{rr}(R) - \sigma_{\theta\theta}(R) \tag{14}
$$

Re-inserting this result into (1) provides the explicit expression for the interface acceleration $\ddot{R}$:

$$
\ddot{R} = \frac{1}{\rho R} \left[ P_g(R) - P_0 - P_A(t) - \frac{2\gamma}{R} + (\sigma_{rr} - \sigma_{\theta\theta}) - \frac{1}{2}\rho \dot{R}^2 \right] \tag{15}
$$

---

## 5. Coupled First-Order ODE System

The non-local integral formulation is thus reduced to a system of four stiff, coupled first-order Ordinary Differential Equations suitable for time-marching integration schemes:

$$
\frac{\mathrm{d}}{\mathrm{d}t} 
\begin{pmatrix} 
R \\[4pt] 
\dot{R} \\[4pt] 
\sigma_{rr} \\[4pt] 
\sigma_{\theta\theta} 
\end{pmatrix} 
= 
\begin{pmatrix} 
\dot{R} \\[4pt]
\frac{1}{\rho R} \left[ P_g(R) - P_0 - P_A(t) - \frac{2\gamma}{R} + (\sigma_{rr} - \sigma_{\theta\theta}) - \frac{1}{2}\rho \dot{R}^2 \right] \\[4pt]
-4G\left(\frac{\dot{R}}{R}\right) - \frac{\sigma_{rr}}{\lambda} - 4\left(\frac{\dot{R}}{R}\right)\sigma_{rr} \\[4pt]
2G\left(\frac{\dot{R}}{R}\right) - \frac{\sigma_{\theta\theta}}{\lambda} + 2\left(\frac{\dot{R}}{R}\right)\sigma_{\theta\theta}
\end{pmatrix} \tag{16}
$$

We first import the dependencies for numerical computation, integration, and visualization. $\textcolor{magenta}{numpy}$ handles vector operations and mathematical constants, $\textcolor{magenta}{\texttt{scipy.integrate.solve\_ivp}}$ provides the $\textcolor{magenta}{adaptive \;initial \;value \;problem\; solver \;Radau}$ used to integrate the differential equations, and $\textcolor{magenta}{matplotlib.pyplot}$ is used to render the dimensionless dynamic response plots.

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp
```

This block sets up the physical fluid properties ($\rho, \gamma, \eta$), acoustic driving parameters ($f = 3\text{ MHz}$, $p_A = 0.4\text{ MPa}$), and initial internal polytropic gas pressure $P_{g0}$. It also establishes the evaluation time grid spanning $9$ acoustic cycles ($9/f$) sampled at $1500$ uniform intervals.

```python
# --- Physical Parameters & Time Grid ---
R0, f, pA = 1.5e-6, 3.0e6, 0.4e6
rho, gamma, eta = 1000.0, 0.072, 0.03
P0, kappa = 101325.0, 1.4
Pg0 = P0 + (2 * gamma / R0)

t_end = 9.0 / f
t_eval = np.linspace(0, t_end, 1500)
```

This subroutine encapsulates the system ODEs, dynamically selecting between the standard 2nd-order Newtonian formulation ($De = 0$) and the stiff 4th-order UCM model ($De > 0$) that couples radial motion with extra-stress rates ($\mathrm{d}\sigma_{rr}/\mathrm{d}t, \mathrm{d}\sigma_{\theta\theta}/\mathrm{d}t$). Integration is performed using the implicit Radau solver to handle numerical stiffness during bubble collapse phases.

```python
def simulate_bubble(De):
    if De == 0:
        # Classical Newtonian viscous Rayleigh-Plesset equation
        def ode(t, y):
            R, Rdot = y
            R_safe = max(R, 1e-12)

            PA = -pA * np.sin(2 * np.pi * f * t)
            Pg = Pg0 * (R0 / R_safe) ** (3 * kappa)

            Rddot = (1 / (rho * R_safe)) * (
                Pg
                - P0
                - PA
                - (2 * gamma / R_safe)
                - (4 * eta * Rdot / R_safe)
                - 0.5 * rho * Rdot**2
            )
            return [Rdot, Rddot]

        y0 = [R0, 0.0]
        sol = solve_ivp(
            ode,
            [0, t_end],
            y0,
            t_eval=t_eval,
            method="Radau",
            rtol=1e-6,
            atol=1e-8,
        )
        return sol.t, sol.y[0]

    else:
        # Upper Convective Maxwell (UCM) viscoelastic model
        lmbda = De / (2 * np.pi * f)
        G = eta / lmbda

        def ode(t, y):
            R, Rdot, tau_rr, tau_tt = y
            R_safe = max(R, 1e-12)
            strain_rate = Rdot / R_safe

            PA = -pA * np.sin(2 * np.pi * f * t)
            Pg = Pg0 * (R0 / R_safe) ** (3 * kappa)

            # Rayleigh-Plesset equation with normal stress difference (tau_rr - tau_tt)
            Rddot = (1 / (rho * R_safe)) * (
                Pg
                - P0
                - PA
                - (2 * gamma / R_safe)
                + (tau_rr - tau_tt)
                - 0.5 * rho * Rdot**2
            )

            # Correct UCM constitutive stress derivatives at bubble boundary r = R(t)
            d_tau_rr = -4.0 * G * strain_rate - (tau_rr / lmbda) - 4.0 * strain_rate * tau_rr
            d_tau_tt =  2.0 * G * strain_rate - (tau_tt / lmbda) + 2.0 * strain_rate * tau_tt

            return [Rdot, Rddot, d_tau_rr, d_tau_tt]

        y0 = [R0, 0.0, 0.0, 0.0]
        sol = solve_ivp(
            ode,
            [0, t_end],
            y0,
            t_eval=t_eval,
            method="Radau",
            rtol=1e-6,
            atol=1e-8,
        )
        return sol.t, sol.y[0]
```

This block will executes a parameter sweep over Deborah numbers ($De \in \{0, 1, 2\}$) to evaluate viscoelastic relaxation effects on cavity dynamics. It normalizes time to acoustic cycles ($ft$) and radius to equilibrium size ($R/R_0$), rendering a comparative transient response plot.

```python
plt.rcParams.update(
    {
        "text.usetex": True,
        "font.family": "serif",
        "font.serif": ["Computer Modern Roman"],
        "xtick.direction": "in",
        "ytick.direction": "in",
        "xtick.top": True,
        "ytick.right": True,
        "xtick.labelsize": 12,
        "ytick.labelsize": 12,
        "axes.linewidth": 1.2,
    }
)

# --- Color Definitions (Dark Line, Lighter Semi-Transparent Marker Face) ---
style_config = {
    0: {"line": "#0B51C1", "face": "#5C96F7"},  # Blue
    1: {"line": "#E00000", "face": "#FF7070"},  # Red
    2: {"line": "#238B45", "face": "#74C476"},  # Green
}

plt.figure(figsize=(4.5, 3.8), dpi=300)

for De in [0, 1, 2]:
    t, R = simulate_bubble(De)
    cycles = f * t  # Dimensionless time ft

    plt.plot(
        cycles,
        R / R0,
        label=rf"$\mathrm{{De}} = {De}$",
        color=style_config[De]["line"],
        linewidth=2.2,
        marker="o",
        markersize=5.5,
        markerfacecolor=style_config[De]["face"],
        markeredgecolor=style_config[De]["line"],
        markeredgewidth=0.6,
        markevery=5,  # Subsamples points to avoid overlapping markers
    )

# --- Labels and Axis Limits ---
plt.xlabel(r"$ft$", fontsize=24, labelpad=8)
plt.ylabel(r"$R/R_0$", fontsize=24, labelpad=8)
plt.xlim(0, 9)
plt.ylim(0, 5)

# --- Legend (Frameless, Match JFM Style) ---
plt.legend(loc="upper left", fontsize=14, frameon=False, handlelength=2.5)

plt.tight_layout()
plt.show()
```

## 1. PINNS formulation technique

The large magnitude of physical stresses ($\tau_{rr}, \tau_{\theta\theta} \sim 10^6 \text{ Pa}$) creates ill-conditioned loss surfaces, exploding gradients, and severe optimization failure when training Physics-Informed Neural Networks (PINNs). Scaling all state variables to $\mathcal{O}(1)$ eliminates this gradient imbalance and speeds up neural network convergence.

### Reference Scales & Non-Dimensional Variables

Using standard characteristic quantities from the system:

* **Length scale:** $R_0 = 1.5 \times 10^{-6}\text{ m} \quad \implies \quad \hat{R} = \frac{R}{R_0}$
* **Time scale:** $t_0 = \frac{1}{f} = \frac{1}{3 \times 10^6}\text{ s} \quad \implies \quad \hat{t} = f t$
* **Velocity scale:** $v_0 = R_0 f = 4.5\text{ m/s} \quad \implies \quad \hat{v} = \frac{\dot{R}}{R_0 f}$
* **Stress / Pressure scale:** $P_0 = 101325\text{ Pa} \quad \implies \quad \hat{\tau}_{rr} = \frac{\tau_{rr}}{P_0}, \quad \hat{\tau}_{\theta\theta} = \frac{\tau_{\theta\theta}}{P_0}$

### Dimensionless ODE System

Applying chain-rule transformations $\frac{\mathrm{d}}{\mathrm{d}t} = f \frac{\mathrm{d}}{\mathrm{d}\hat{t}}$ yields the scaled system:

$$
\frac{\mathrm{d}}{\mathrm{d}\hat{t}} 
\begin{pmatrix} 
\hat{R} \\ 
\hat{v} \\ 
\hat{\tau}_{rr} \\ 
\hat{\tau}_{\theta\theta} 
\end{pmatrix} 
= 
\begin{pmatrix} 
\hat{v} \\[6pt]
\frac{1}{\mathcal{I} \hat{R}} \left[ \hat{P}_{g0} \hat{R}^{-3\kappa} - 1 + \hat{P}_A \sin(2\pi \hat{t}) - \frac{\hat{\gamma}}{\hat{R}} + (\hat{\tau}_{rr} - \hat{\tau}_{\theta\theta}) \right] - \frac{\hat{v}^2}{2\hat{R}} \\[6pt]
-4\hat{G}\left(\frac{\hat{v}}{\hat{R}}\right) - \frac{\hat{\tau}_{rr}}{\hat{\lambda}} - 4\left(\frac{\hat{v}}{\hat{R}}\right)\hat{\tau}_{rr} \\[6pt]
2\hat{G}\left(\frac{\hat{v}}{\hat{R}}\right) - \frac{\hat{\tau}_{\theta\theta}}{\hat{\lambda}} + 2\left(\frac{\hat{v}}{\hat{R}}\right)\hat{\tau}_{\theta\theta} 
\end{pmatrix}
$$

### Dimensionless Control Parameters

| Parameter | Symbol | Definition / Formula | Value / Estimate |
| :--- | :---: | :---: | :---: |
| **Inertia Parameter** | $\mathcal{I}$ | $\frac{\rho R_0^2 f^2}{P_0}$ | $\approx 0.19985$ |
| **Acoustic Pressure Ratio** | $\hat{P}_A$ | $\frac{p_A}{P_0}$ | $\approx 3.9477$ |
| **Surface Tension Parameter** | $\hat{\gamma}$ | $\frac{2\gamma}{P_0 R_0}$ | $\approx 0.9474$ |
| **Initial Gas Pressure Ratio** | $\hat{P}_{g0}$ | $1 + \hat{\gamma}$ | $\approx 1.9474$ |
| **Relaxation Time Parameter** | $\hat{\lambda}$ | $\lambda f = \frac{\text{De}}{2\pi}$ | — |
| **Elastic Modulus Ratio** | $\hat{G}$ | $\frac{G}{P_0} = \frac{\eta \cdot 2\pi f}{\text{De} \cdot P_0}$ | $\approx \frac{5.5809}{\text{De}}$ |

### Dimensionless Initial Conditions

| State Variable | Dimensional Initial Condition | Dimensionless Initial Condition |
| :--- | :---: | :---: |
| **Radius ($R$)** | $R(0) = R_0$ | $\hat{R}(0) = \frac{R_0}{R_0} = \mathbf{1.0}$ |
| **Velocity ($\dot{R}$)** | $\dot{R}(0) = 0\text{ m/s}$ | $\hat{v}(0) = \frac{0}{R_0 f} = \mathbf{0.0}$ |
| **Radial Stress ($\tau_{rr}$)** | $\tau_{rr}(0) = 0\text{ Pa}$ | $\hat{\tau}_{rr}(0) = \frac{0}{P_0} = \mathbf{0.0}$ |
| **Azimuthal Stress ($\tau_{\theta\theta}$)** | $\tau_{\theta\theta}(0) = 0\text{ Pa}$ | $\hat{\tau}_{\theta\theta}(0) = \frac{0}{P_0} = \mathbf{0.0}$ |

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# --- Dimensionless Physical Constants ---
R0, f, pA = 1.5e-6, 3.0e6, 0.4e6
rho, gamma, eta = 1000.0, 0.072, 0.03
P0, kappa = 101325.0, 1.4

# Scaling Ratios
P_A_hat = pA / P0
gamma_hat = (2.0 * gamma) / (P0 * R0)
P_g0_hat = 1.0 + gamma_hat
I_param = (rho * (R0**2) * (f**2)) / P0

t_end_hat = 9.0
t_eval_hat = np.linspace(0, t_end_hat, 1500)


def simulate_nondim(De):
    if De == 0:
        # Newtonian case
        def ode(t_hat, y):
            R_hat, v_hat = y
            R_safe = max(R_hat, 1e-12)

            PA_hat = -P_A_hat * np.sin(2.0 * np.pi * t_hat)
            Pg_hat = P_g0_hat * (1.0 / R_safe) ** (3.0 * kappa)
            eta_hat = (eta * f) / P0

            v_dot_hat = (1.0 / (I_param * R_safe)) * (
                Pg_hat
                - 1.0
                - PA_hat
                - (gamma_hat / R_safe)
                - (4.0 * eta_hat * v_hat / R_safe)
            ) - 0.5 * (v_hat**2 / R_safe)

            return [v_hat, v_dot_hat]

        sol = solve_ivp(
            ode,
            [0, t_end_hat],
            [1.0, 0.0],
            t_eval=t_eval_hat,
            method="Radau",
            rtol=1e-6,
            atol=1e-8,
        )
        return sol.t, sol.y[0], sol.y[1], np.zeros_like(sol.t), np.zeros_like(sol.t)

    else:
        # Viscoelastic Upper Convective Maxwell case
        lambda_hat = De / (2.0 * np.pi)
        G_hat = (eta * 2.0 * np.pi * f / De) / P0

        def ode(t_hat, y):
            R_hat, v_hat, tau_rr_hat, tau_tt_hat = y
            R_safe = max(R_hat, 1e-12)
            sr = v_hat / R_safe  # Strain rate

            PA_hat = -P_A_hat * np.sin(2.0 * np.pi * t_hat)
            Pg_hat = P_g0_hat * (1.0 / R_safe) ** (3.0 * kappa)

            v_dot_hat = (1.0 / (I_param * R_safe)) * (
                Pg_hat
                - 1.0
                - PA_hat
                - (gamma_hat / R_safe)
                + (tau_rr_hat - tau_tt_hat)
            ) - 0.5 * (v_hat**2 / R_safe)

            d_tau_rr_hat = (
                -4.0 * G_hat * sr - (tau_rr_hat / lambda_hat) - 4.0 * sr * tau_rr_hat
            )
            d_tau_tt_hat = (
                2.0 * G_hat * sr - (tau_tt_hat / lambda_hat) + 2.0 * sr * tau_tt_hat
            )

            return [v_hat, v_dot_hat, d_tau_rr_hat, d_tau_tt_hat]

        sol = solve_ivp(
            ode,
            [0, t_end_hat],
            [1.0, 0.0, 0.0, 0.0],
            t_eval=t_eval_hat,
            method="Radau",
            rtol=1e-6,
            atol=1e-8,
        )
        return sol.t, sol.y[0], sol.y[1], sol.y[2], sol.y[3]


# --- Global Matplotlib Style Settings ---
plt.rcParams.update(
    {
        "text.usetex": True,
        "font.family": "serif",
        "font.serif": ["Computer Modern Roman"],
        "xtick.direction": "in",
        "ytick.direction": "in",
        "xtick.top": True,
        "ytick.right": True,
        "xtick.labelsize": 11,
        "ytick.labelsize": 11,
        "axes.linewidth": 1.1,
    }
)

style_config = {
    0: {"line": "#0B51C1", "face": "#5C96F7"},  # Blue
    1: {"line": "#E00000", "face": "#FF7070"},  # Red
    2: {"line": "#238B45", "face": "#74C476"},  # Green
}

# Precompute data for De = 0, 1, 2
data = {}
for De in [0, 1, 2]:
    t_hat, R_hat, v_hat, tau_rr_hat, tau_tt_hat = simulate_nondim(De)
    data[De] = {
        "t": t_hat,
        "R": R_hat,
        "v": v_hat,
        "tau_rr": tau_rr_hat,
        "tau_tt": tau_tt_hat,
    }


# ==============================================================================
# FIGURE 1: 4-SUBPLOT BREAKDOWN (R_hat, v_hat, tau_rr_hat, tau_tt_hat)
# ==============================================================================
fig, axes = plt.subplots(2, 2, figsize=(9, 7), dpi=300)

for De in [0, 1, 2]:
    t = data[De]["t"]
    line_color = style_config[De]["line"]

    # (a) Dimensionless Radius
    axes[0, 0].plot(
        t, data[De]["R"], label=rf"$\mathrm{{De}} = {De}$", color=line_color, lw=1.8
    )

    # (b) Dimensionless Velocity
    axes[0, 1].plot(t, data[De]["v"], color=line_color, lw=1.8)

    # (c) Dimensionless Radial Stress
    axes[1, 0].plot(t, data[De]["tau_rr"], color=line_color, lw=1.8)

    # (d) Dimensionless Azimuthal Stress
    axes[1, 1].plot(t, data[De]["tau_tt"], color=line_color, lw=1.8)

# --- Subplot Formatting & Axis Labels ---
axes[0, 0].set_ylabel(r"$\hat{R} = R/R_0$", fontsize=14)
axes[0, 0].set_xlim(0, 9)
axes[0, 0].grid(True, linestyle="--", alpha=0.4)
axes[0, 0].legend(loc="upper left", fontsize=10, frameon=False)
axes[0, 0].text(
    0.03, 0.92, r"\textbf{(a)}", transform=axes[0, 0].transAxes, fontsize=12
)

axes[0, 1].set_ylabel(r"$\hat{v} = \dot{R}/(R_0 f)$", fontsize=14)
axes[0, 1].set_xlim(0, 9)
axes[0, 1].grid(True, linestyle="--", alpha=0.4)
axes[0, 1].text(
    0.03, 0.92, r"\textbf{(b)}", transform=axes[0, 1].transAxes, fontsize=12
)

axes[1, 0].set_xlabel(r"$\hat{t} = ft$", fontsize=14)
axes[1, 0].set_ylabel(r"$\hat{\tau}_{rr} = \tau_{rr}/P_0$", fontsize=14)
axes[1, 0].set_xlim(0, 9)
axes[1, 0].grid(True, linestyle="--", alpha=0.4)
axes[1, 0].text(
    0.03, 0.92, r"\textbf{(c)}", transform=axes[1, 0].transAxes, fontsize=12
)

axes[1, 1].set_xlabel(r"$\hat{t} = ft$", fontsize=14)
axes[1, 1].set_ylabel(r"$\hat{\tau}_{\theta\theta} = \tau_{\theta\theta}/P_0$", fontsize=14)
axes[1, 1].set_xlim(0, 9)
axes[1, 1].grid(True, linestyle="--", alpha=0.4)
axes[1, 1].text(
    0.03, 0.92, r"\textbf{(d)}", transform=axes[1, 1].transAxes, fontsize=12
)

plt.tight_layout()
plt.show()
```
